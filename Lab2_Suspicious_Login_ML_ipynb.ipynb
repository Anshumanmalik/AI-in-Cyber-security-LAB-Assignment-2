# LAB 2: Detecting Suspicious Login Activity Using Machine Learning

**ENSP355 | AI in Cyber Security Lab | 2026-27** — Total: 50 marks

Problem: numeric login-activity data (failed logins, login hour, new-device flag,
downloaded MB) ko clean karke Decision Tree se Normal/Suspicious classify karna,
K-Means se cluster karna, aur results interpret karna.

> Colab me chalane se pehle `simple_login_activity.csv` upload karein (left Files panel → Upload).


```python
# Setup: CSV load
import pandas as pd
import numpy as np

df = pd.read_csv('simple_login_activity.csv')
df.head()
```

## Q1. Dataset Inspection and Cleaning (5 marks)

**Aim:** input columns aur target label pehchanna, duplicates hatana, missing numeric values bharna.

- Models ko complete, consistent data chahiye; median missing numeric values ke liye robust hai.
- Duplicate rows analysis ko distort karte hain.
- Target label: `Normal` / `Suspicious`; baaki selected columns input features hain.


```python
print(df.info())
print('Missing values before cleaning:')
print(df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())

df = df.drop_duplicates().copy()
df['failed_logins'] = df['failed_logins'].fillna(df['failed_logins'].median())
df['download_mb'] = df['download_mb'].fillna(df['download_mb'].median())

print('Missing values after cleaning:', df.isna().sum().sum())
print('Duplicate rows after cleaning:', df.duplicated().sum())
print('Final rows:', len(df))
```

## Q2. Training and Testing Split (5 marks)

**Aim:** features/label alag karke 75/25 train-test split banana.

- Training data se model seekhta hai; testing data unseen records pe performance batata hai.
- Stratified split dono sets me class proportion same rakhta hai.
- `random_state=42` se split reproducible hai.


```python
from sklearn.model_selection import train_test_split

features = ['failed_logins', 'login_hour', 'new_device', 'download_mb']
X = df[features]
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)

print('Training rows:', len(X_train))
print('Testing rows:', len(X_test))
```

## Q3. Decision Tree Training and Evaluation (10 marks)

**Aim:** Decision Tree train karke accuracy aur confusion matrix interpret karna.

- Tree labelled data se if-then rules seekhta hai; `max_depth=4` tree ko readable
  rakhta hai aur overfitting kam karta hai.
- Confusion matrix: sahi Normal/Suspicious faisle vs false alarms (FP) aur missed attacks (FN).


```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

model = DecisionTreeClassifier(max_depth=4, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy*100:.2f}%')

ConfusionMatrixDisplay.from_predictions(y_test, y_pred,
    labels=['Normal', 'Suspicious'])
plt.title('Decision Tree Confusion Matrix')
plt.show()
```

## Q4. K-Means Clustering (10 marks)

**Aim:** unsupervised tareeke se similar login activity group karna aur labels se compare karna.

- K-Means labels use nahi karta, sirf numeric similarity pe group banata hai.
- Alag scales (hour vs MB) ki wajah se features ko standardize karna zaroori hai.
- Clusters ka labels se perfect match hona zaroori nahi.


```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X_scaled)

cluster_summary = df.groupby('cluster')[features].mean().round(2)
label_summary = pd.crosstab(df['cluster'], df['label'])
print(cluster_summary.to_string())
print()
print(label_summary.to_string())
```

## Q5. Feature Importance and New Record Prediction (5 marks)

**Aim:** tree ne kin features ko sabse zyada use kiya, aur 3 naye records predict karna.

- Feature importance batata hai tree ne split ke liye kis feature ko kitna use kiya.
- Ye model-specific hai; causation prove nahi karta.


```python
importance = pd.DataFrame({
    'feature': features,
    'importance': model.feature_importances_
}).sort_values('importance', ascending=False)
print(importance.to_string(index=False))

plt.bar(importance['feature'], importance['importance'])
plt.title('Decision Tree Feature Importance')
plt.ylabel('Importance')
plt.xticks(rotation=25)
plt.tight_layout()
plt.show()

new_records = pd.DataFrame([
    {'failed_logins': 0, 'login_hour': 10, 'new_device': 0, 'download_mb': 45},
    {'failed_logins': 8, 'login_hour': 2,  'new_device': 1, 'download_mb': 180},
    {'failed_logins': 2, 'login_hour': 23, 'new_device': 1, 'download_mb': 900},
])
new_records['prediction'] = model.predict(new_records)
print(new_records.to_string(index=False))
```

## Q6. False Positives, False Negatives and Reflection (5 marks)

**Aim:** do zaroori model errors samjhana aur model ki suitability summarize karna.

- **False positive:** genuine activity galat tareeke se Suspicious mark ho jaye (analyst ka time waste).
- **False negative:** asli attack Normal mark ho jaye (attack miss ho sakta hai).
- Test set me 0 false positives mile; neeche ke 3 false negatives model ke **asli** misses hain.


```python
print('False positive example:',
      'Legitimate late-night work on a new device is marked Suspicious.')
print('False negative example:',
      'A disguised attack with few obvious indicators is marked Normal.')

# test set me asli misses nikaalte hain
test_check = X_test.copy()
test_check['actual'] = y_test.values
test_check['predicted'] = y_pred
fns = test_check[(test_check['actual'] == 'Suspicious') &
                 (test_check['predicted'] == 'Normal')]
print('Real false negatives in the test set:')
print(fns.to_string())

df.to_csv('Lab2_Login_ML_Result.csv', index=False)
print('Saved: Lab2_Login_ML_Result.csv')
```
